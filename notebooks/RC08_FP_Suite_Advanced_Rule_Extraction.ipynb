# Ultra-Optimized FP-Growth Association Rule Pipeline

## Introduction

### Executive Brief
This project implements a **high-performance FP-Growth association rule mining pipeline** using `mlxtend.frequent_patterns`, designed for **large-scale transactional datasets (1M+ rows)**. The pipeline converts transactional data (`user_id`, `product_id`) into encoded baskets, then applies **FP-Growth**, **FPMax**, and **Association Rules** to discover meaningful relationships between items. The architecture is optimized for production environments using **joblib parallel processing**, **tqdm progress bars**, **global logging**, and robust **error handling**, ensuring scalability, traceability, and efficient computation.

The algorithm flow begins with **Transaction Encoding**, where transactions are grouped by `user_id` and transformed into a one-hot encoded matrix. Next, **FP-Growth** extracts frequent itemsets based on configurable `min_support`, followed by **FPMax** to identify maximal frequent patterns. These itemsets are then passed into the **Association Rule Generator**, which calculates metrics such as **support**, **confidence**, and **lift**. Finally, a **parallel prediction engine** evaluates transactions against learned rules to generate recommendation candidates efficiently across multiple CPU cores.

In conclusion, this pipeline provides a **production-ready association rule engine** capable of handling large-scale recommender systems with high performance and flexibility. With modular components, global logging, and parallel execution, the architecture supports future enhancements such as **incremental learning**, **streaming FP-Growth**, **DuckDB integration**, and **Cython acceleration**, making it suitable for enterprise-level recommendation systems and advanced market basket analysis.

### Mathematical Logic & Formulation
#### Transaction Definition
Let:
* Users
  $U = {u_1, u_2, … , u_m}$
* Items
  $I = {i_1, i_1, … , i_m}$
* Transactions
  $T = {t_1, t_2, … , t_k}$

Each transaction $t_k \subseteq I$

---

### Frequent Itemset Mining
#### Support
Support measures how often an itemset appears:
$$\text{Support}(X) = \frac{\text{Count}(X)}{N}$$
**Where:**
* $X$ = itemset
* $N$ = total transactions

#### Association Rules
**Rule:**
$$X \implies Y$$
**Where:**
* $X$ = antecedent
* $Y$ = consequent

#### Confidence
Confidence measures rule accuracy:
$$\text{Confidence}(X \implies Y) = \frac{\text{Support}(X \cup Y)}{\text{Support}(X)}$$

#### Lift
Lift measures correlation strength:
$$\text{Lift}(X \implies Y) = \frac{\text{Support}(X \cup Y)}{\text{Support}(X) \times \text{Support}(Y)}$$
Interpretation:
* $\text{Lift} > 1 \implies$ Positive correlation
* $\text{Lift} = 1 \implies$ Independent
* $\text{Lift} < 1 \implies$ Negative correlation

#### Leverage
$$\text{Leverage}(X \implies Y) = \text{Support}(X \cup Y) - (\text{Support}(X) \times \text{Support}(Y))$$

#### Conviction
$$\text{Conviction}(X \implies Y) = \frac{1 - \text{Support}(Y)}{1 - \text{Confidence}(X \implies Y)}$$

```python
import duckdb
import pandas as pd
from typing import Optional, Union, List, Dict, Any, ContextManager
from pathlib import Path
from contextlib import contextmanager

class DuckDBManager:
    """
    Manager class for DuckDB connections and queries.

    Example:
        db = DuckDBManager('mydb.duckdb')
        df = db.query("SELECT * FROM users")
        db.close()

        # Or use context manager:
        with DuckDBManager('mydb.duckdb') as db:
            df = db.query("SELECT * FROM users")
    """

    def __init__(self,
                 db_path: Union[str, Path],
                 read_only: bool = True,
                 threads: Optional[int] = None,
                 memory_limit: Optional[str] = '2GB'):
        """
        Initialize DuckDB manager.

        Args:
            db_path: Path to database file (use ':memory:' for in-memory DB)
            read_only: Open in read-only mode
            threads: Number of CPU threads to use
            memory_limit: Memory limit (e.g., '4GB', '1TB')
        """
        self.db_path = str(db_path)
        self.read_only = read_only
        self.conn = None

        # Create connection
        self._create_connection(threads, memory_limit)

    def _create_connection(self, threads: Optional[int] = None,
                          memory_limit: Optional[str] = None):
        """Create DuckDB connection with optional settings."""
        self.conn = duckdb.connect(database=self.db_path, read_only=self.read_only)

        # Configure settings
        if threads:
            self.conn.execute(f"SET threads = {threads}")
        if memory_limit:
            self.conn.execute(f"SET memory_limit = '{memory_limit}'")

    def query(self,
              query: str,
              params: Optional[Union[List, Dict, tuple]] = None,
              fetch_size: Optional[int] = None) -> pd.DataFrame:
        """
        Execute query and return DataFrame.

        Args:
            query: SQL query string
            params: Query parameters
            fetch_size: Number of rows to fetch (None for all)

        Returns:
            pandas DataFrame
        """
        if self.conn is None:
            raise ValueError("Connection is closed. Please reconnect.")

        try:
            if params is not None:
                result = self.conn.execute(query, params)
            else:
                result = self.conn.execute(query)

            if fetch_size:
                return result.fetch_df_chunk(fetch_size)
            return result.fetchdf()

        except Exception as e:
            raise Exception(f"Query failed: {e}\nQuery: {query}")

    def query_arrow(self, query: str, params: Optional[Union[List, Dict]] = None):
        """Execute query and return Arrow table (faster for large datasets)."""
        if params is not None:
            return self.conn.execute(query, params).fetch_arrow_table()
        return self.conn.execute(query).fetch_arrow_table()

    def register_dataframe(self, name: str, df: pd.DataFrame):
        """Register pandas DataFrame as temporary table (table_view)."""
        self.conn.register(name, df)

    def ListedTable(self):
        tables = self.conn.execute("SELECT table_name FROM duckdb_tables()").df()
        data = list()
        for table in tables['table_name']:
            count = self.conn.execute(f"SELECT count(*) FROM {table}").fetchone()[0]
            data.append({'table_name': table, 'row_count': count})
        info_df = pd.DataFrame(data)
        info_df = info_df.sort_values(by='row_count', ascending=False)
        info_df['row_count'] = info_df['row_count'].apply(lambda x: f"{x:,}")
        display(info_df)
        return info_df

    def execute(self, query: str, params: Optional[Union[List, Dict]] = None):
        """Execute query without returning results."""
        if params is not None:
            self.conn.execute(query, params)
        else:
            self.conn.execute(query)

    def table_exists(self, table_name: str) -> bool:
        """Check if table exists in database."""
        result = self.query(
            "SELECT COUNT(*) FROM information_schema.tables WHERE table_name = ?",
            [table_name]
        )
        return result.iloc[0, 0] > 0

    def get_tables(self) -> List[str]:
        """Get list of all tables in database."""
        df = self.query("SELECT table_name FROM information_schema.tables WHERE table_schema = 'main'")
        return df['table_name'].tolist()

    def get_schema(self, table_name: str) -> pd.DataFrame:
        """Get schema information for a table."""
        return self.query(f"DESCRIBE {table_name}")

    def close(self):
        """Close database connection."""
        if self.conn:
            self.conn.close()
            self.conn = None

    def __enter__(self):
        """Context manager entry."""
        return self

    def __exit__(self, exc_type, exc_val, exc_tb):
        """Context manager exit."""
        self.close()
```

```python
@contextmanager
def duckdb_connection(db_path: Union[str, Path],
                      read_only: bool = False,
                      **kwargs) -> DuckDBManager:
    """
    Context manager for DuckDB connection.

    Example:
        with duckdb_connection('mydb.duckdb') as db:
            df = db.query("SELECT * FROM users")
    """
    db = DuckDBManager(db_path, read_only, **kwargs)
    try:
        yield db
    finally:
        db.close()
```

## Basic Function and Decorator

```python
logging.basicConfig(
    level=logging.DEBUG,
    format="%(asctime)s | %(levelname)s | %(message)s"
)

logger = logging.getLogger("FP-GROWTH")
```

```python
@contextmanager
def joblib_parallel(n_jobs:int=-1,
                    backend:str="loky",
                    verbose:int=10,
                    total:int=None,
                    desc:str="Processing",
                   ):
    """
    Advanced Joblib Context Manager.
    Optimized for memory-heavy MLXtend operations.
    Sample usage:
    tasks = [delayed(my_prediction_func)(data) for data in batch_list]
    with joblib_parallel(n_jobs=-1, total=len(tasks), desc="RC08 Mining") as (parallel, pbar):
        results = parallel(tasks)
    """
    actual_jobs = os.cpu_count() if n_jobs == -1 else n_jobs
    logger.debug(f"Initializing Parallel Engine | Backend: {backend} | Workers: {actual_jobs}")
    pbar = tqdm(total=total, desc=desc, colour='green', disable=not total)
    try:
        with parallel_config(backend=backend, n_jobs=n_jobs, verbose=verbose):
            def tqdm_parallel(iterable):
                for item in iterable:
                    yield item
                    pbar.update(1)

            yield Parallel(n_jobs=n_jobs, backend = backend), pbar

    except Exception as arc:
        logger.error(f"Parallel Execution Error: {str(arc)}")
        raise RuntimeError(arc)
    finally:
        logger.debug("Cleaning up Parallel workers and memory.")
        gc.collect()
        if backend == "loky":
            from joblib.externals.loky import get_reusable_executor
            get_reusable_executor().shutdown(wait=True)
        logger.info("Parallel Processing Finished")
```

```python
from typing import Callable, List, Any, TypeVar, Iterable

T = TypeVar('T')  # Input data type
R = TypeVar('R')  # Return data type

def The_PTasks(
         task_func    : Callable[[T], R],
         data_batches : Iterable[T],
         n_jobs       : int = -1,
         desc         : str = "RC08 Parallel Task"
    ) -> List[R]:
    """
    Orchestrates the parallel execution of tasks with strict typing and
    multi-layered error handling. For example:
    results: List[dict] = execute_parallel_tasks(
                            task_func    = my_prediction_func,
                            data_batches = user_baskets,
                            desc         = "Rule Prediction Engine",)
    """
    if not data_batches:
        logger.warning(f"Empty data passed to {desc}. Returning empty list.")
        return list()

    try:
        # Create the delayed task list
        # Typing: task_func takes a 'T' from data_batches and returns an 'R'
        tasks = [delayed(task_func)(batch) for batch in data_batches]
        with joblib_parallel(n_jobs=n_jobs, total=len(tasks), desc=desc) as (parallel, pbar):
            def pbar_wrapper(iterable):
                for task in iterable:
                    yield task
            results = parallel(tasks)
        return results

    except TypeError as te:
        logger.error(f"Signature Mismatch in '{desc}': {te}")
        raise TypeError(f"The task_func provided is not compatible with the data in data_batches: {te}")

    except MemoryError as me:
        logger.error(f"Memory Exhausted during '{desc}': {me}")
        gc.collect()
        raise MemoryError(f"RC08 Parallel Engine ran out of RAM. try reducing n_jobs: {me}")

    except Exception as e:
        logger.error(f"Unexpected Failure in '{desc}': {type(e).__name__} - {e}")
        raise RuntimeError(f"RC08 Parallel execution failed unexpectedly: {e}") from e
```

```python
import traceback

def log_step(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        logger.debug(f"START {func.__name__}")
        try:
            result = func(*args, **kwargs)
            logger.debug(f"END {func.__name__}")
            return result
        except Exception as e:

            logger.error(f"ERROR in {func.__name__}")
            logger.error(str(e))
            logger.error(traceback.format_exc())
            raise
    return wrapper
```

## Transaction Encoder

```python
class TransactionEncoderModel:
    def __init__(self):
        self._transactions = None
        self._encoded = None

    @property
    def transactions(self):
        return self._transactions

    @property
    def encoded(self):
        return self._encoded

    @log_step
    def fit_transform(
        self,
        df,
        user_col="user_id",
        item_col="product_id"
    ):
        logger.debug("Grouping Transactions")
        grouped = (
            df.groupby(user_col)[item_col]
            .apply(list))
        self._transactions = grouped.tolist()
        logger.debug(f"Total Transactions {len(self._transactions)}")

        logger.debug("Transaction Encoding")
        te = TransactionEncoder()
        encoded = te.fit(self._transactions).transform(self._transactions)
        self._encoded = pd.DataFrame(encoded, columns=te.columns_)
        logger.debug(f"Encoded Shape {self._encoded.shape}")
        return self._encoded
```

## FP Growth Model

```python
class FPGrowthModel:
    def __init__(
        self,
        min_support=0.001,
    ):
        self._min_support = min_support
        self.frequent_itemsets = None

    @property
    def min_support(self):
        return self._min_support

    @min_support.setter
    def min_support(self, value):
        logger.debug(f"Set min_support {value}")
        self._min_support = value

    @log_step
    def fit(self, X):
        logger.debug("Running FP‑Growth")
        self.frequent_itemsets = fpgrowth(
            X,
            min_support=self.min_support,
            use_colnames=True)
        logger.debug(f"Frequent Itemsets {self.frequent_itemsets.shape}")
        return self.frequent_itemsets
```

## FPMax Model

```python
class FPMaxModel:
    def __init__(self, min_support=0.001):
        self.min_support = min_support
        self.frequent_itemsets = None

    @log_step
    def fit(self, X):
        logger.debug("Running FPMax")
        self.frequent_itemsets = fpmax(
            X,
            min_support=self.min_support,
            use_colnames=True
        )
        return self.frequent_itemsets
```

## Association Rules

```python
class AssociationRuleModel:
    def __init__(
        self,
        metric="lift",
        threshold=1.0
    ):
        self.metric = metric
        self.threshold = threshold

    @log_step
    def fit(self, frequent_itemsets):
        logger.debug("Generating Association Rules")
        rules = association_rules(
            frequent_itemsets,
            metric=self.metric,
            min_threshold=self.threshold
        )
        logger.debug(f"Rules Generated {rules.shape}")
        return rules
```

## Rule Predictor

```python
from functools import partial

class RulePredictor:
    def __init__(self, n_jobs=-1):
        self.n_jobs = n_jobs

    def _predict_row(self, basket, rules):
        basket = set(basket)
        matches = rules[
            rules['antecedents'].apply(
                lambda x: x.issubset(basket)
            )]
        return matches

    @log_step
    def predict(self, transactions, rules):
        logger.debug("Predicting Rules")
        results = np.ndarray([])
        try:
            task_to_run = partial(self._predict_row, rules=rules)
            results = The_PTasks(
                task_func    = task_to_run,   # Mapping: Callable[[T], R]
                data_batches = transactions,  # Mapping: Iterable[T]
                n_jobs       = self.n_jobs,   # Mapping: int
                desc         = "Predicting"   # Mapping: str
            )
        except Exception as arch:
            logger.error(f"Runtime error : {arch}")
            with joblib_parallel(self.n_jobs) as parallel:
                results = parallel(
                    delayed(self._predict_row)(
                        t, rules)
                    for t in tqdm(transactions, desc="Predicting")
                )
        finally:
            return results
```

## Master Pipeline

```python
class FPPipeline:
    def __init__(
        self,
        min_support=0.001,
        metric="lift",
        threshold=1.0,
        n_jobs=-1
    ):
        self.encoder = TransactionEncoderModel()
        self.fpgrowth = FPGrowthModel(
            min_support=min_support)
        self.fpmax = FPMaxModel(min_support=min_support)
        self.rules_model = AssociationRuleModel(
            metric=metric,
            threshold=threshold)
        self.predictor = RulePredictor(n_jobs=n_jobs)

    @log_step
    def fit(self, df):
        logger.debug("Encoding Data")
        X = self.encoder.fit_transform(df)
        logger.debug("Training FP‑Growth")

        fp_growth = self.fpgrowth.fit(X)
        logger.debug("Training FPMax")

        fp_max = self.fpmax.fit(X)
        logger.debug("Training Association Rules")

        rules = self.rules_model.fit(fp_growth)
        self.rules = rules
        return {
            "fpgrowth": fp_growth,
            "fpmax": fp_max,
            "rules": rules,
        }

    @log_step
    def predict(self):
        transactions = self.encoder.transactions
        return self.predictor.predict(
            transactions,
            self.rules)
```

## API Function

```python
@log_step
def run_fp_pipeline(
    data : pd.DataFrame,
    min_support=0.001,
    metric="lift",
    threshold=1.0,
    n_jobs=-1
):
    pipeline = FPPipeline(
        min_support=min_support,
        metric=metric,
        threshold=threshold,
        n_jobs=n_jobs
    )
    results = pipeline.fit(data)
    predictions = pipeline.predict()
    return pipeline, results, predictions
```
